```python
def ffdls(items, bin_sizes):

    def try_shift(bin_idx, bins, bin_sizes, max_capacity):
        cap, space, contents = bins[bin_idx]
        if any(i > (1/3) * max_capacity for i in contents):
            bin_used = cap - space
            candidate_bins = [c for c in bin_sizes if c < cap and c >= (3/4) * cap]
            if candidate_bins:
                smallest_fit = min(candidate_bins)
                if bin_used <= smallest_fit:
                    bins[bin_idx] = (smallest_fit, smallest_fit - bin_used, contents.copy())

    items = sorted(items, reverse=True)
    bin_sizes = sorted(bin_sizes, reverse=True)
    max_capacity = bin_sizes[0]

    bins = []  

    for item in items:
        placed = False
        for idx, (cap, space, contents) in enumerate(bins):
            if space >= item:
                contents = contents.copy()
                contents.append(item)
                bins[idx] = (cap, space - item, contents)

                try_shift(idx, bins, bin_sizes, max_capacity)
                placed = True
                break

        if not placed:
            bins.append((max_capacity, max_capacity - item, [item]))

    all_items = []
    for _, _, contents in bins:
        all_items.extend(contents)
    all_items = sorted(all_items, reverse=True)

    final_bins = []
    bin_sizes_sorted = sorted(bin_sizes)  

    for item in all_items:
        best_fit = -1
        min_space_left = float('inf')
        for idx, (cap, space, contents) in enumerate(final_bins):
            if space >= item and (space - item) < min_space_left:
                best_fit = idx
                min_space_left = space - item

        if best_fit != -1:
            cap, space, contents = final_bins[best_fit]
            contents = contents.copy()
            contents.append(item)
            final_bins[best_fit] = (cap, space - item, contents)
        else:
            for cap in bin_sizes_sorted:
                if cap >= item:
                    final_bins.append((cap, cap - item, [item]))
                    break
            else:
                raise Exception(f"Item {item} no cabe en ningún bin disponible")

    return [(cap, cap - space, contents) for cap, space, contents in final_bins]


def print_bin_summary(bins):
    print("Resultado del FFDLS:\n")
    for i, (cap, used, contents) in enumerate(bins):
        print(f"Bin {i + 1}: Capacidad={cap}, Usado={used}, Contenido={contents}, "
              f"Ocupación={round(used / cap * 100, 1)}%")
    total_cost = sum(cap for cap, _, _ in bins)
    print(f"\nCosto total (suma de capacidades usadas): {total_cost}")
```

```python
import random

def generar_instancia_para_ffdls(num_items, item_range, bin_capacities, seed=None):
    if seed is not None:
        random.seed(seed)

    items = [random.randint(*item_range) for _ in range(num_items)]
    return items, bin_capacities

items, bin_capacities = generar_instancia_para_ffdls(
    num_items=100,
    item_range=(1, 13),
    bin_capacities=[10, 30, 50],
    seed=42
)

print("Items generados:", items)
print("Capacidades de bins:", bin_capacities)

resultado = ffdls(items, bin_capacities)
print_bin_summary(resultado)
```
